```python
# Bayes Theorem

# Prior probability of having the disease
P_D = 1 / 1000

# Probability of not having the disease
P_not_D = 1 - P_D

# Probability of a positive test when disease is present
P_Pos_D = 0.95

# False positive probability
P_Pos_not_D = 0.01

# Calculate total probability of a positive test
P_Pos = (P_Pos_D * P_D) + (P_Pos_not_D * P_not_D)

# Apply Bayes theorem
P_D_Pos = (P_Pos_D * P_D) / P_Pos

# Display result
print("Probability of disease given a positive test:",
      f"{P_D_Pos:.2f}")

print("Percentage probability:",
      f"{P_D_Pos * 100:.2f}", "%")
```
